```python
import pandas as pd 
import matplotlib.pyplot as plt

df = pd.read_excel('Starbucks1.xlsx', header=None, names=['review'])

def sentiment_analysis(text):
    if pd.isnull(text):  
        return 0
    if 'super delicious' in text or 'excellent' in text or 'best' in text or 'lovely' in text or 'friendly' in text or 'great' in text or 'nice' in text or 'delicious' in text or 'immaculate' in text or 'polite' in text or 'attentive' in text or 'prompt' in text or 'helpful' in text or 'cosy' in text or 'up to scratch' in text or 'fine' in text:
        return 1 
    elif 'awful' in text or 'stale' in text or 'rude' in text or 'terrible' in text or 'grim' in text or 'scalding' in text or 'not fresh' in text or 'omg' in text:
        return -1 
    else:
        return 0  


df['sentiment'] = df['review'].apply(sentiment_analysis)


sentiment_counts = df['sentiment'].value_counts().sort_index()


plt.figure(figsize=(10, 6))
bars = plt.bar(['Negative', 'Neutral', 'Positive'], sentiment_counts.values, color=['red', 'blue', 'green'])


for bar, count in zip(bars, sentiment_counts.values):
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2 - 0.1, yval + 0.5, count, ha='center', va='bottom')


plt.title('Sentiment Analysis of Starbucks Reviews')
plt.xlabel('Sentiment')
plt.ylabel('Number of Reviews')

plt.show()
```
